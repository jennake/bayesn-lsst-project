```python
import numpyro

numpyro.set_host_device_count(4)

import jax
import jax.numpy as jnp
import numpy as np
import matplotlib.pyplot as plt

from bayesn import SEDmodel
```

```python
print(jax.devices())
print("Number of devices:", jax.local_device_count())
```

```python
filt_map = {'g': 'g_PS1','r': 'r_PS1','i': 'i_PS1','z': 'z_PS1'}

z_err_values = [0.005, 0.01, 0.02, 0.05, 0.1] # err increments instead
photoz_results = {}

for z_err in z_err_values:
    print(f"Running z_prior_err={z_err}")

    fit_model = SEDmodel(load_model='T21_model')

    samples, props = fit_model.fit_from_file(
        fit_model.example_lc,
        filt_map=filt_map,
        photoz=True,
        z_prior_err=z_err,
        chain_method='parallel' # change to parallel?
    )

    photoz_results[z_err] = {
        'samples': samples,
        'props': props
    }
```

```python
filt_map = {'g': 'g_PS1','r': 'r_PS1','i': 'i_PS1','z': 'z_PS1'}

z_err_values = [0.005, 0.01, 0.02, 0.05, 0.1] # err increments instead
photoz_results = {}

for z_err in z_err_values:
    print(f"Running z_prior_err={z_err}")

    fit_model = SEDmodel(load_model='T21_model')

    samples, props = fit_model.fit_from_file(
        fit_model.example_lc,
        filt_map=filt_map,
        photoz=True,
        z_prior_err=z_err,
        chain_method='parallel' # change to parallel?
    )

    photoz_results[z_err] = {
        'samples': samples,
        'props': props
    }
```

```python
for z_err, result in photoz_results.items():
    z_samples = np.asarray(result['samples']['z']).reshape(-1)
    z16, z50, z84 = np.percentile(z_samples, [16, 50, 84])

    print(
        f"prior sigma={z_err:.3f}: "
        f"median={z50:.5f}, "
        f"68% CI=[{z16:.5f}, {z84:.5f}], "
        f"posterior SD={z_samples.std():.5f}"
    )
```

```python
fixed_z = 0.019253

for z_err, result in photoz_results.items():
    z_samples = np.asarray(result['samples']['z']).reshape(-1)

    plt.hist(
        z_samples,
        bins=40,
        density=True,
        histtype='step',
        label=f'prior σ={z_err}'
    )

plt.axvline(
    fixed_z,
    linestyle='--',
    label='Input redshift'
)

plt.xlabel('Redshift')
plt.ylabel('Posterior density')
plt.legend()
plt.show()
```

```python
posterior_sd = []

for z_err in z_err_values:
    z_samples = np.asarray(
        photoz_results[z_err]['samples']['z']
    ).reshape(-1)

    posterior_sd.append(z_samples.std())

plt.plot(z_err_values, posterior_sd, marker='o')
plt.plot(z_err_values, z_err_values, linestyle='--')

plt.xlabel('Prior redshift standard deviation')
plt.ylabel('Posterior redshift standard deviation')
plt.show()
```
